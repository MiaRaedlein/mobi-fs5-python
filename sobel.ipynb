**[Task 2.2.]()** Implement the two functions `sobel_h` and `sobel_v` for horizontal and vertical $3 \times 3$ Sobel filtering, respectively:

```python
from scipy.signal import convolve2d

def sobel_h(img):
    sobel_h = (1/8) * np.array([
        [1, 0, -1],
        [2, 0, -2],
        [1, 0, -1]
    ])
    g_x = convolve2d(img, sobel_h, mode='same', boundary='symm')
    return g_x
    
def sobel_v(img):
    sobel_v = (1/8) * np.array([
            [1, 2, 1],
            [0, 0, 0],
            [-1, -2, -1]
        ])
    g_y = convolve2d(img, sobel_v, mode='same', boundary='symm')
    return g_y
```

**[Task 2.3.]()** Test your implementations from above:
