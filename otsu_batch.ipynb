**[Task 2.1.]()** Implement Otsu thresholding as a re-usable function:

```python
def otsu_thresholding(imgf):
    img8 = (imgf * 255).round().astype(np.uint8)


    h = np.zeros(256)

    for i in range(256) :
        h[i] = np.sum(img8 == i)

    j = np.zeros(256)

    for T in range(256):
        j[T] = (np.sum(h[:T])*np.var(img8[img8 < T]) + (np.sum(h[T:])*np.var(img8[img8 >= T])))

    T_best = np.nanargmin(j)

    img8.shape

    img_2 = img8.copy()
    T = T_best

    for x in range(1024) :
        for y in range(1344) : 
            if img_2[x,y] >= T:
                img_2[x,y] = 1
            else :
                img_2[x,y] = 0
    
    return img_2
```

**[Task 2.2.]()** Implement a re-usable function to compute the Dice coefficient:

```python
def compute_dice(bin1, bin2):
    assert bin1.dtype == bool
    assert bin2.dtype == bool    

    G = bin1

    H = bin2

    GH = G * H

    DC = 2*GH.sum()/(G.sum() + H.sum())

    
    return DC 
```
